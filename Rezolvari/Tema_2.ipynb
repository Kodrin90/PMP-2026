## Exercițiul 1 — Aruncarea a două zaruri

Spațiul evenimentelor elementare este $\Omega = \{1, 2, 3, 4, 5, 6\}^2$, cu $|\Omega| = 36$ perechi $(X, Y)$ echiprobabile.

1. **Probabilitatea ca suma să fie cel puțin 10, știind că primul zar indică un număr par:**
   * $B = \{X \text{ este par}\} = \{2, 4, 6\} \times \{1, \dots, 6\} \implies |B| = 18$.
   * $A = \{X + Y \ge 10\} = \{(4,6), (5,5), (5,6), (6,4), (6,5), (6,6)\} \implies |A| = 6$.
   * $A \cap B = \{(4,6), (6,4), (6,5), (6,6)\} \implies |A \cap B| = 4$.
   * $P(A \mid B) = \frac{|A \cap B|}{|B|} = \frac{4}{18} = \frac{2}{9} \approx 0.222222$.

2. **Probabilitatea ca suma să fie cel puțin 10, știind că cel puțin unul dintre zaruri indică un număr par:**
   * $C = \{\text{cel puțin un zar este par}\} = \Omega \setminus \{X \text{ impar și } Y \text{ impar}\} \implies |C| = 36 - 9 = 27$.
   * $A \cap C = A \setminus \{(5,5)\} = \{(4,6), (5,6), (6,4), (6,5), (6,6)\} \implies |A \cap C| = 5$.
   * $P(A \mid C) = \frac{|A \cap C|}{|C|} = \frac{5}{27} \approx 0.185185$.

3. **Probabilitatea ca ambele zaruri să indice numere pare, știind că suma lor este 8:**
   * $D = \{X + Y = 8\} = \{(2,6), (3,5), (4,4), (5,3), (6,2)\} \implies |D| = 5$.
   * $E = \{X \text{ par și } Y \text{ par}\} \implies E \cap D = \{(2,6), (4,4), (6,2)\} \implies |E \cap D| = 3$.
   * $P(E \mid D) = \frac{|E \cap D|}{|D|} = \frac{3}{5} = 0.600000$.

4. **Sunt evenimentele $A = \{X + Y \ge 10\}$ și $B = \{X \text{ este par}\}$ independente?**
   * **Nu**, evenimentele $A$ și $B$ **nu sunt independente**, deoarece:
     $$P(A \mid B) = \frac{2}{9} \approx 0.2222 \neq P(A) = \frac{6}{36} = \frac{1}{6} \approx 0.1667$$
     (sau echivalent, $P(A \cap B) = \frac{4}{36} = \frac{1}{9} \neq P(A) \cdot P(B) = \frac{1}{6} \cdot \frac{1}{2} = \frac{1}{12}$).

```python
import numpy as np

# Extensie Python: Simularea a 100.000 de aruncări și comparația cu valorile exacte
rng = np.random.default_rng(42)
N = 100_000

X = rng.integers(1, 7, size=N)
Y = rng.integers(1, 7, size=N)

A = (X + Y) >= 10
B = (X % 2) == 0
C = ((X % 2) == 0) | ((Y % 2) == 0)
D = (X + Y) == 8
E = ((X % 2) == 0) & ((Y % 2) == 0)

p1_sim = np.mean(A[B])
p2_sim = np.mean(A[C])
p3_sim = np.mean(E[D])

print(f"1. P(X+Y >= 10 | X par):               simulat = {p1_sim:.5f} | exact = {2/9:.5f}")
print(f"2. P(X+Y >= 10 | cel puțin un par):    simulat = {p2_sim:.5f} | exact = {5/27:.5f}")
print(f"3. P(ambele pare | X+Y = 8):           simulat = {p3_sim:.5f} | exact = {3/5:.5f}")
print(f"4. Verificare independență A și B:     P(A∩B) ≈ {np.mean(A & B):.5f} vs P(A)P(B) ≈ {np.mean(A)*np.mean(B):.5f}")
```

## Exercițiul 2 — Filtrarea mesajelor spam

Notăm evenimentele:
* $S$: mesajul este spam ($P(S) = 0.30 \implies P(\neg S) = 0.70$);
* $O$: mesajul conține cuvântul „ofertă” ($P(O \mid S) = 0.80, \; P(O \mid \neg S) = 0.15$);
* $G$: mesajul conține termenul „gratuit” ($P(G \mid S) = 0.60, \; P(G \mid \neg S) = 0.10$).

1. **Probabilitatea ca un mesaj care conține „ofertă” să fie spam ($P(S \mid O)$):**
   $$P(O) = P(O \mid S)P(S) + P(O \mid \neg S)P(\neg S) = 0.80 \times 0.30 + 0.15 \times 0.70 = 0.24 + 0.105 = 0.345.$$
   $$P(S \mid O) = \frac{P(O \mid S)P(S)}{P(O)} = \frac{0.24}{0.345} = \frac{16}{23} \approx 0.695652 \quad (69.57\%).$$

2. **Probabilitatea ca un mesaj care conține ambele cuvinte („ofertă” și „gratuit”) să fie spam ($P(S \mid O \cap G)$):**
   Presupunând independența condiționată a cuvintelor dată fiind clasa mesajului:
   $$P(O \cap G \mid S) = P(O \mid S) P(G \mid S) = 0.80 \times 0.60 = 0.48,$$
   $$P(O \cap G \mid \neg S) = P(O \mid \neg S) P(G \mid \neg S) = 0.15 \times 0.10 = 0.015.$$
   Prin formula lui Bayes:
   $$P(O \cap G) = 0.48 \times 0.30 + 0.015 \times 0.70 = 0.144 + 0.0105 = 0.1545,$$
   $$P(S \mid O \cap G) = \frac{0.144}{0.1545} = \frac{96}{103} \approx 0.932039 \quad (93.20\%).$$

3. **De ce ipoteza independenței condiționate poate fi nerealistă?**
   În limbajul natural, cuvintele nu apar independent nici măcar în interiorul aceleiași categorii de mesaje. Termenii *„ofertă”* și *„gratuit”* au o puternică corelație semantică și frazeologică (ex. sintagme frecvente precum *„ofertă gratuită”* sau *„încearcă gratuit oferta noastră”*), astfel încât $P(G \mid O, S) > P(G \mid S)$ și $P(G \mid O, \neg S) > P(G \mid \neg S)$. Ignorarea acestei dependențe poate distorsiona estimarea probabilității a posteriori (fenomen specific clasificatorului Naive Bayes).

4. **Regulă de clasificare bazată pe probabilitatea condiționată:**
   Pentru un mesaj caracterizat de vectorul de atribute observate $\mathbf{x}$, folosim regula de decizie Bayes cu un prag $\tau \in (0, 1)$:
   $$\text{Clasificare}(\mathbf{x}) = \begin{cases} \text{Spam}, & \text{dacă } P(S \mid \mathbf{x}) \ge \tau, \\ \text{Legitim}, & \text{dacă } P(S \mid \mathbf{x}) < \tau. \end{cases}$$
   * Pentru costuri egale ale erorilor (regula MAP standard), se alege $\tau = 0.5$.
   * În practică, deoarece blocarea unui e-mail legitim important (fals pozitiv) este mai costisitoare decât primirea unui spam în Inbox (fals negativ), se alege un prag mai ridicat, de exemplu $\tau = 0.80$ sau $\tau = 0.90$ (echivalent cu $\log \frac{P(S)}{P(\neg S)} + \sum_i \log \frac{P(x_i \mid S)}{P(x_i \mid \neg S)} \ge \log \frac{\tau}{1 - \tau}$).

```python
# Verificare numerică pentru Exercițiul 2
P_S = 0.30
P_not_S = 0.70
P_O_S, P_O_not_S = 0.80, 0.15
P_G_S, P_G_not_S = 0.60, 0.10

P_S_given_O = (P_O_S * P_S) / (P_O_S * P_S + P_O_not_S * P_not_S)
P_S_given_OG = (P_O_S * P_G_S * P_S) / (P_O_S * P_G_S * P_S + P_O_not_S * P_G_not_S * P_not_S)

print(f"P(Spam | 'ofertă')              = {P_S_given_O:.6f} ({P_S_given_O*100:.2f}%)")
print(f"P(Spam | 'ofertă' și 'gratuit') = {P_S_given_OG:.6f} ({P_S_given_OG*100:.2f}%)")
```

## Exercițiul 3 — Paradoxul lui Simpson

1. **Rata de promovare pe fiecare program:**
   * **Programul 1:**
     * Disciplina A: $\frac{81}{90} = 0.90 = 90\%$.
     * Disciplina B: $\frac{19}{20} = 0.95 = 95\%$.
   * **Programul 2:**
     * Disciplina A: $\frac{1}{10} = 0.10 = 10\%$.
     * Disciplina B: $\frac{16}{80} = 0.20 = 20\%$.

2. **Care disciplină are o rată de promovare mai mare în fiecare program?**
   În **ambele programe**, **Disciplina B** are o rată de promovare mai ridicată ($95\% > 90\%$ în Programul 1 și $20\% > 10\%$ în Programul 2).

3. **Rata globală de promovare (date agregate):**
   * **Disciplina A:** $\frac{81 + 1}{90 + 10} = \frac{82}{100} = 82\%$.
   * **Disciplina B:** $\frac{19 + 16}{20 + 80} = \frac{35}{100} = 35\%$.
   La nivel agregat, ordinea se inversează: **Disciplina A** pare mult mai ușoară decât Disciplina B ($82\% > 35\%$).

4. **Explicația fenomenului:**
   Programul de studiu este o **variabilă de confuzie (confounder)**. Studenții din Programul 1 au rate foarte mari de promovare la ambele discipline ($90\%$ și $95\%$), în timp ce studenții din Programul 2 au rate foarte scăzute ($10\%$ și $20\%$). Deoarece la Disciplina A au fost evaluați preponderent studenți din Programul 1 ($90$ din $100$), iar la Disciplina B preponderent studenți din Programul 2 ($80$ din $100$), agregarea ignoră dimensiunile inegale ale subgrupurilor și răstoarnă comparația.

5. **Formulare în termeni de probabilități condiționate:**
   Fie $R = \{\text{Promovat}\}$, $A, B$ disciplinele și $P_1, P_2$ programele de studiu. Paradoxul lui Simpson constă în faptul că:
   $$P(R \mid B, P_1) > P(R \mid A, P_1) \quad \text{și} \quad P(R \mid B, P_2) > P(R \mid A, P_2),$$
   însă prin formula probabilității totale:
   $$P(R \mid A) = P(R \mid A, P_1)P(P_1 \mid A) + P(R \mid A, P_2)P(P_2 \mid A) = 0.90 \times 0.90 + 0.10 \times 0.10 = 0.82,$$
   $$P(R \mid B) = P(R \mid B, P_1)P(P_1 \mid B) + P(R \mid B, P_2)P(P_2 \mid B) = 0.95 \times 0.20 + 0.20 \times 0.80 = 0.35,$$
   ceea ce conduce la $P(R \mid A) > P(R \mid B)$ din cauza ponderilor condiționate opuse $P(P_1 \mid A) = 0.90$ față de $P(P_1 \mid B) = 0.20$.

```python
import pandas as pd

# 1. Verificarea datelor din problemă cu pandas
df_univ = pd.DataFrame([
    {"Program": "Programul 1", "Disciplina": "A", "Promovati": 81, "Total": 90},
    {"Program": "Programul 1", "Disciplina": "B", "Promovati": 19, "Total": 20},
    {"Program": "Programul 2", "Disciplina": "A", "Promovati": 1,  "Total": 10},
    {"Program": "Programul 2", "Disciplina": "B", "Promovati": 16, "Total": 80},
])
df_univ["Rata_promovare (%)"] = (df_univ["Promovati"] / df_univ["Total"]) * 100
print("=== Datele problemei pe subgrupuri ===")
print(df_univ.to_string(index=False))

agg_univ = df_univ.groupby("Disciplina", as_index=False)[["Promovati", "Total"]].sum()
agg_univ["Rata_globala (%)"] = (agg_univ["Promovati"] / agg_univ["Total"]) * 100
print("\n=== Datele agregate global (inversarea ordinii!) ===")
print(agg_univ.to_string(index=False))

# 2. Exemplu propriu de Paradox Simpson: Eficiența a două tratamente medicale (T1 vs T2)
# Confounder: Severitatea cazului (Ușor vs Sever)
df_medical = pd.DataFrame([
    {"Severitate": "Cazuri Ușoare", "Tratament": "Tratament T1", "Vindecati": 92, "Pacienti": 100},  # 92%
    {"Severitate": "Cazuri Ușoare", "Tratament": "Tratament T2", "Vindecati": 19, "Pacienti": 20},   # 95%
    {"Severitate": "Cazuri Severe", "Tratament": "Tratament T1", "Vindecati": 12, "Pacienti": 40},   # 30%
    {"Severitate": "Cazuri Severe", "Tratament": "Tratament T2", "Vindecati": 70, "Pacienti": 200},  # 35%
])
df_medical["Rata_vindecare (%)"] = (df_medical["Vindecati"] / df_medical["Pacienti"]) * 100

print("\n=== Exemplu propriu (Tratament medical) — Pe categorii de severitate ===")
print(df_medical.to_string(index=False))

agg_med = df_medical.groupby("Tratament", as_index=False)[["Vindecati", "Pacienti"]].sum()
agg_med["Rata_globala (%)"] = (agg_med["Vindecati"] / agg_med["Pacienti"]) * 100
print("\n=== Exemplu propriu — Agregat global (T1 pare mai bun: 74.29% vs 40.45%, deși T2 e mai bun în ambele grupe!) ===")
print(agg_med.to_string(index=False))
```
